```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np

model = LinearRegression()
data = np.array([1,2,3,4,5,6])
y = np.array([1,4,9,16,25,36])
plt.scatter(data, y)
plt.grid()
plt.show()
```

```python
# Reshape data to 2D array as required by scikit-learn
X = data.reshape(-1, 1)

# Train (fit) the model
model.fit(X, y)

# Make predictions using the trained model
y_pred = model.predict(X)

# Visualize the original data and the trained model's regression line
plt.scatter(data, y, color='green', label='Actual Data')
plt.plot(data, y_pred, color='yellow', linewidth=2, label='Linear Regression Line')
plt.xlabel('X (data)')
plt.ylabel('y')
plt.title('Simple Linear Regression Training')
plt.legend()
plt.grid(True)
plt.show()

# Print the model parameters
print(f"Model Coefficient (Slope): {model.coef_[0]:.4f}")
print(f"Model Intercept: {model.intercept_:.4f}")
```

```python
from sklearn.preprocessing import PolynomialFeatures

# Définir le degré souhaité (par exemple, 2 pour une relation quadratique)
degree = 2
poly_features = PolynomialFeatures(degree=degree, include_bias=False)

# Transformer les caractéristiques d'origine pour inclure les puissances supérieures
X_poly = poly_features.fit_transform(X)

# Créer et entraîner un nouveau modèle de régression linéaire sur ces caractéristiques transformées
model_poly = LinearRegression()
model_poly.fit(X_poly, y)

# Prédire sur un intervalle continu plus fluide pour un meilleur rendu visuel
X_fit = np.linspace(1, 6, 100).reshape(-1, 1)
X_fit_poly = poly_features.transform(X_fit)
y_fit = model_poly.predict(X_fit_poly)

# Visualisation des résultats
plt.scatter(data, y, color='green', label='Données réelles')
plt.plot(X_fit, y_fit, color='red', linewidth=2, label=f'Régression Polynomiale (degré {degree})')
plt.xlabel('X (data)')
plt.ylabel('y')
plt.title(f'Régression Polynomiale de Degré {degree}')
plt.legend()
plt.grid(True)
plt.show()

# Affichage des coefficients
print(f"Coefficients du modèle : {model_poly.coef_}")
print(f"Constante (Intercept) : {model_poly.intercept_:.4f}")
```

```python
from sklearn.pipeline import make_pipeline

# Création d'un pipeline qui combine la transformation polynomiale et la régression linéaire
degree = 2
pipeline_model = make_pipeline(
    PolynomialFeatures(degree=degree, include_bias=False),
    LinearRegression()
)

# Entraînement du pipeline complet directement sur les données d'origine X
pipeline_model.fit(X, y)

# Prédiction sur l'intervalle continu pour l'affichage visuel
X_fit = np.linspace(1, 6, 100).reshape(-1, 1)
y_fit_pipe = pipeline_model.predict(X_fit)

# Visualisation des résultats
print(pipeline_model.predict([[10]]))
plt.scatter(data, y, color='green', label='Données réelles')
plt.plot(X_fit, y_fit_pipe, color='magenta', linewidth=2, label=f'Pipeline (degré {degree})')
plt.xlabel('X (data)')
plt.ylabel('y')
plt.title('Régression Polynomiale via Pipeline')
plt.legend()
plt.grid(True)
plt.show()

# Pour accéder aux coefficients individuels à l'intérieur du pipeline
linear_reg_step = pipeline_model.named_steps['linearregression']
print(f"Coefficients du modèle : {linear_reg_step.coef_}")
print(f"Constante (Intercept) : {linear_reg_step.intercept_:.4f}")
```
